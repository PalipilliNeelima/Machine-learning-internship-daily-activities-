Feature Engineering:
------------------------------
Feature engineering is the process of converting raw data into meaningful numerical features so that a machine learning model can understand patterns and make better predictions.

1.Feature Exploration:
--------------------------------
The process of analyzing and visualizing features to uncover patterns, distributions, correlations, and potential issues in the dataset before applying machine learning.

2.Feature Creation:
----------------------------
The process of generating new meaningful features from existing data using domain knowledge or mathemaatical operations to help the model learn better patterns.

3.Feature Encoding:
-----------------------------
The process of converting categorical or non-eric features into numeric form so machine learning models can process them effectively.

4.Feature Scaling:(normalization- standard ,minmax scaler)
--------------------------
The process of transforming numerical features to a standard range or distribution so that no feature dominates the learning process due to its scale.

5.Feature Selection:
-----------------------------
The process of identifying and keeping only the most relevant features while removing irrelevant or redundant ones too improve model performance and reduce overfitting.

# Feature Creation


```python
import pandas as pd
df=pd.DataFrame({
    "study_hours":[2,4,6,8,10],
    "attendance":[60,70,80,90,95],
    "maths":[40,55,65,75,80],
    "science":[42,58,68,78,88]
})
df
```

# Domain- Based Feature

```python
df["total_marks"]=df["maths"]+df["science"]
df
```

# Mathematical Feature

```python
df["marks_per_hour"]=df["total_marks"] / df["study_hours"]
df
```

# Interaction Feature

```python
df["study_attendance_interaction"]= df["study_hours"] * df["attendance"]
df
```

# Polynomial Feature

```python
df["study_hours_squared"]= df["study_hours"] ** 2
df
```

 Feature Selection:
 ---------------------------
It is the process of selecting only relevant feeatures and removing irrelevant or redundant ones to improve model performance.

why it "Remove Noise":
-----------------------------------
- Too many features confuse the model 
- Irrelevant features cause overfitting
- Noise reduces prediction accuracy

Types of Feature Selection:
----------------------------------------
- Filter methods(correlation, chi-square)
- Wrapper Methods(forward, backward, RFE)
- Embedded methods(Lasso, three-based models)

Example:
--------------
- removing weakly correlated features
- dropping features with zero importance
- selecting top features using RFE

Impact on model:
-------------------------
- reduces overfitting
- improves generalization
- speeds up training
