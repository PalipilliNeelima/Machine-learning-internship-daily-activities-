# IQR Method

```python
q1=data.quantile(0.25)
q3=data.quantile(0.75)
IQR = q3-q1
LB=q1- 1.5*IQR
UB=q3+ 1.5*IQR
outliers=data[(data<(LB))|(data>(UB))]
```

# z-score

```python
import pandas as pd
from scipy import stats
import numpy as np
import matplotlib.pyplot as plt

data=pd.DataFrame({
    "study_hours":[1,2,3,4,5,6,7,8,9,20]
})
data  
```

```python
z_scores=np.abs(stats.zscore(data["study_hours"]))
outliers= data[z_scores>2]
outliers
```

```python
plt.scatter(range(len(data)),data["study_hours"])
plt.xlabel("Index")
plt.ylabel("Study Hours")
plt.title("Scatter Plot showing outlier")
plt.show()
```

```python
plt.hist(data["study_hours"], bins=40)
plt.xlabel("Index")
plt.ylabel("Study Hours")
plt.title("Histogram showing outlier")
plt.show()
```
