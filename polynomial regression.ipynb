# Polynomial regression
- polynomial regression is a regression technique used to model a non-linear relationship between an independent variable and a dependent varaible by expressing the input variable as a ploynomial of degree greater than one, while still using linear regression to estimate the coefficients.

# general polynomial regression equation


```python
import pandas as pd
df=pd.DataFrame({
    "study_hours":[1,2,3,4,5],
    "marks":[10,25,55,90,130]
})
```

```python
df
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
data={
    "study_hours":[1,2,3,4,5],
    "marks":[10,25,55,90,130]
}
df=pd.DataFrame(data)
df
```

```python
x=df[["study_hours"]]
y=df["marks"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
poly=PolynomialFeatures(degree=2)
x_train_poly=poly.fit_transform(x_train)
x_test_poly=poly.transform(x_test)
```

```python
model=LinearRegression()
model.fit(x_train_poly,y_train)
```

```python
y_train_pred=model.predict(x_train_poly)
y_test_pred=model.predict(x_test_poly)
```

```python
print("training mse:",mean_squared_error(y_train,y_train_pred))
print("testing mse:",mean_squared_error(y_test,y_test_pred))
```

```python
plt.scatter(x,y,color="blue",label="actual data")
x_line=np.linspace(min(x['study_hours']),max(x["study_hours"]),100).reshape(-1,1)
y_line=model.predict(poly.transform(x_line))
plt.plot(x_line,y_line,color="red",label="polynomial fit")
plt.xlabel("study_hours")
plt.ylabel("marks")
plt.title("polynomial regression example")
plt.legend()
plt.show()
```

```python
#linear regression
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
# predictiions
y_train_pred=model.predict(x_train)
y_test_pred=model.predict(x_test)
```

```python
#  evaluate
print("training mse:",mean_squared_error(y_train,y_train_pred))
print("testing mse:",mean_squared_error(y_test,y_test_pred))
```

```python
#plot
plt.scatter(x,y,color="blue",label="actual data")
plt.plot(x,model.predict(x),color="red",label="linear-fit")
plt.xlabel("study hours")
plt.ylabel("marks")
plt.title("linear regression example")
plt.legend()
plt.show()
```
